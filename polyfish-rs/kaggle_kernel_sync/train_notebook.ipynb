```python
!pip install safetensors torch
import os
import shutil
dataset_path = None
for root, dirs, files in os.walk('/kaggle/input'):
    if 'train.py' in files:
        dataset_path = root
        break
if not dataset_path:
    raise FileNotFoundError('Could not find train.py in /kaggle/input')
print('Found dataset path:', dataset_path)
working_dir = '/kaggle/working'
# Copy dataset to working dir so train.py can write/read properly
for item in os.listdir(dataset_path):
    s = os.path.join(dataset_path, item)
    d = os.path.join(working_dir, item)
    if os.path.isdir(s):
        shutil.copytree(s, d, dirs_exist_ok=True)
    else:
        shutil.copy2(s, d)
```
